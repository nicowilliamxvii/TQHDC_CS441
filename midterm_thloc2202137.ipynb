  5.1 Visualization 1 – Top & Bottom Happiest Countries
Problem

Mục tiêu của phần này là xác định những quốc gia có mức độ hạnh phúc cao nhất và thấp nhất dựa trên biến Happiness Score.
Câu hỏi đặt ra là: Quốc gia nào hạnh phúc nhất và quốc gia nào kém hạnh phúc nhất?

**  Data**

Dữ liệu được sử dụng từ bộ World Happiness Report, trong đó biến chính được phân tích là:

Happiness Score – điểm số thể hiện mức độ hạnh phúc của mỗi quốc gia

Country – tên quốc gia

Computation and Visualization

Để trả lời câu hỏi nghiên cứu, các bước xử lý dữ liệu được thực hiện như sau:

Sắp xếp dữ liệu theo biến Happiness Score theo thứ tự giảm dần

Chọn 10 quốc gia có điểm cao nhất (top 10)

Chọn 10 quốc gia có điểm thấp nhất (bottom 10)

Sử dụng biểu đồ cột ngang (horizontal bar chart) để trực quan hóa dữ liệu

Hình 1: Top 10 và Bottom 10 quốc gia theo điểm hạnh phúc

Biểu đồ cho thấy sự khác biệt rõ rệt giữa hai nhóm quốc gia. Các quốc gia thuộc nhóm top 10 có điểm hạnh phúc cao (thường trên 7), trong khi nhóm bottom 10 có điểm thấp hơn đáng kể (thường dưới 4).

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Sắp xếp dữ liệu theo Happiness Score
sorted_df = df.sort_values(by="Happiness Score", ascending=False)

# Chọn 10 quốc gia hạnh phúc nhất
top10 = sorted_df.head(10)

# Chọn 10 quốc gia kém hạnh phúc nhất
bottom10 = sorted_df.tail(10)

# thêm label để phân biệt
top10["Group"] = "Top 10"
bottom10["Group"] = "Bottom 10"

combined = pd.concat([top10, bottom10])

# sắp xếp lại cho đẹp
combined = combined.sort_values(by="Happiness Score", ascending=True)

plt.figure(figsize=(10,8))

sns.barplot(
    x="Happiness Score",
    y="Country",
    hue="Group",
    data=combined
)

plt.title("Top 10 và Bottom 10 Quốc Gia Hạnh Phúc")
plt.xlabel("Happiness Score")
plt.ylabel("Country")

plt.show()
```

Problem

Câu hỏi đặt ra là: Điểm hạnh phúc tập trung ở khoảng nào và có phân bố đồng đều hay không?

Data

Dữ liệu sử dụng:

Happiness Score – điểm hạnh phúc của các quốc gia

Computation and Visualization

Các bước thực hiện:

Chọn cột Happiness Score

Chia dữ liệu thành các khoảng giá trị

Sử dụng biểu đồ histogram để thể hiện phân bố

Hình 2: Phân bố điểm hạnh phúc

Biểu đồ cho thấy phần lớn các quốc gia có điểm hạnh phúc nằm trong khoảng từ 4 đến 6. Số lượng quốc gia có điểm rất cao hoặc rất thấp là khá ít.

Conclusion

5.3 Visualization 3 – Correlation Between Variables
Problem

Mục tiêu là xác định các yếu tố có ảnh hưởng đến mức độ hạnh phúc.
Câu hỏi đặt ra là: Những biến nào có mối liên hệ mạnh với Happiness Score?

Data

Dữ liệu sử dụng gồm các biến:

Happiness Score

Economy

Health

Family

Freedom

Generosity

Corruption

Computation and Visualization

Các bước thực hiện:

Tính toán ma trận tương quan giữa các biến

Sử dụng heatmap để trực quan hóa

Hình 3: Ma trận tương quan

Biểu đồ cho thấy các biến như GDP (Economy), Health và Social Support (Family) có tương quan dương mạnh với Happiness Score.

Conclusion

Các yếu tố kinh tế, sức khỏe và hỗ trợ xã hội có ảnh hưởng lớn đến mức độ hạnh phúc. Đây là những yếu tố quan trọng nhất cần được xem xét.

```python
# @title
plt.figure(figsize=(10,8))

# Compute correlation matrix
corr_matrix = df.corr(numeric_only=True)

# Plot heatmap
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Between Variables")

plt.show()
```

Computation and Visualization

5.5 Visualization 5 – Regional Happiness Comparison
Problem

Mục tiêu là so sánh mức độ hạnh phúc giữa các khu vực.
Câu hỏi đặt ra là: Các khu vực trên thế giới có sự khác biệt về hạnh phúc không?

Data

Dữ liệu sử dụng:

Happiness Score

Region

Computation and Visualization

Các bước thực hiện:

Nhóm dữ liệu theo Region

Vẽ boxplot để so sánh

Hình 5: So sánh hạnh phúc theo khu vực

Biểu đồ cho thấy một số khu vực có mức hạnh phúc cao hơn, trong khi các khu vực khác có mức thấp hơn và phân bố rộng hơn.

Conclusion

Có sự khác biệt rõ rệt giữa các khu vực. Điều này phản ánh sự khác nhau về điều kiện kinh tế, xã hội và môi trường sống.

```python
plt.figure(figsize=(12,6))

sns.boxplot(
    x="Region",
    y="Happiness Score",
    data=df
)

plt.xticks(rotation=45)

plt.title("Happiness Score by Region")
plt.xlabel("Region")
plt.ylabel("Happiness Score")

plt.show()
```
