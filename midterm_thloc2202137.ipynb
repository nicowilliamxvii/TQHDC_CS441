5.1 Visualization 1 – Top & Bottom Happiest Countries
Goal

The goal of this visualization is to identify the happiest and least happy countries based on the Happiness Score.

Steps

Sort the dataset by Happiness Score.

Select the top 10 happiest countries.

Select the bottom 10 least happy countries.

Create a horizontal bar chart to visualize the scores.

```python
# Sort dataset by happiness score
df_sorted = df.sort_values(by="Happiness Score", ascending=False)
# Top 10 happiest countries
top10 = df_sorted.head(10)
# Bottom 10 least happy countries
bottom10 = df_sorted.tail(10)
```

```python
# Plot top 10
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(10,6))
sns.barplot(
    x="Happiness Score",
    y="Country",
    data=top10,
    palette="viridis"

)

plt.title("Top 10 Happiest Countries")
plt.xlabel("Happiness Score")
plt.ylabel("Country")

plt.show()
```

Quan sát

Biểu đồ cho thấy các quốc gia như Na Uy, Đan Mạch và Iceland nằm trong số những quốc gia hạnh phúc nhất với điểm số hạnh phúc trên 7.

Giải thích

Những quốc gia này nhìn chung có nền kinh tế mạnh, hệ thống hỗ trợ xã hội tốt và hệ thống chăm sóc sức khỏe phát triển. Điều này cho thấy sự ổn định kinh tế và cấu trúc xã hội vững mạnh đóng góp đáng kể vào hạnh phúc.

5.2 Visualization 2 – Happiness Score Distribution
Goal

The purpose of this visualization is to understand how happiness scores are distributed across all countries.

Steps

Select the Happiness Score column.

Divide the values into several ranges using a histogram.

Plot the histogram to observe how scores are distributed.

```python
plt.figure(figsize=(8,6))

sns.histplot(
    df["Happiness Score"],
    bins=20,
    kde=True
)

plt.title("Distribution of Happiness Scores")
plt.xlabel("Happiness Score")
plt.ylabel("Number of Countries")

plt.show()
```

Quan sát

Hầu hết các quốc gia có chỉ số hạnh phúc nằm trong khoảng từ 4 đến 6.

Chỉ một số ít quốc gia có mức độ hạnh phúc cực cao, trên 7.

Giải thích

Điều này cho thấy mức độ hạnh phúc toàn cầu có xu hướng tập trung quanh các giá trị trung bình hơn là các giá trị cực đoan.

5.3 Visualization 3 – Correlation Heatmap
Goal

The objective of this visualization is to analyze the relationships between happiness and other socio-economic variables.

Steps

Select all numerical variables in the dataset.

Compute the correlation matrix.

Display the correlation values using a heatmap.

```python
plt.figure(figsize=(10,8))

# Compute correlation matrix
corr_matrix = df.corr(numeric_only=True)

# Plot heatmap
sns.heatmap(
    corr_matrix,
    annot=True,
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Correlation Between Variables")

plt.show()
```

Quan sát

Biểu đồ nhiệt cho thấy một số biến số có mối tương quan tích cực mạnh mẽ với hạnh phúc, bao gồm:

Kinh tế (GDP bình quân đầu người)

Sức khỏe (Tuổi thọ)

Gia đình (Hỗ trợ xã hội)

Giải thích

Những yếu tố này dường như có ảnh hưởng mạnh nhất đến mức độ hạnh phúc ở các quốc gia.

5.4 Visualization 4 – GDP vs Happiness Score
Goal

This visualization explores the relationship between economic prosperity and happiness.

Steps

Select Economy (GDP per capita) as the x-axis.

Select Happiness Score as the y-axis.

Plot each country as a point in a scatter plot.

```python
plt.figure(figsize=(8,6))

sns.scatterplot(
    x="Economy",
    y="Happiness Score",
    data=df
)

plt.title("GDP per Capita vs Happiness Score")
plt.xlabel("GDP per Capita")
plt.ylabel("Happiness Score")

plt.show()
```

Quan sát

Biểu đồ phân tán cho thấy xu hướng chung là tăng lên, cho thấy các quốc gia có GDP bình quân đầu người cao hơn thường có điểm hạnh phúc cao hơn.

Giải thích

Mặc dù sự thịnh vượng kinh tế có mối liên hệ chặt chẽ với hạnh phúc, nhưng mối quan hệ này không hoàn toàn tuyến tính. Điều này cho thấy các yếu tố khác như sức khỏe, tự do và hỗ trợ xã hội cũng đóng vai trò quan trọng.

5.5 Visualization 5 – Happiness Score by Region
Goal

The purpose of this visualization is to compare happiness levels across different regions of the world.

Steps

Group the dataset by the Region column.

Compare the happiness scores for each region.

Use a box plot to visualize the differences.

```python
plt.figure(figsize=(12,6))

sns.boxplot(
    x="Region",
    y="Happiness Score",
    data=df
)

plt.xticks(rotation=45)

plt.title("Happiness Score by Region")
plt.xlabel("Region")
plt.ylabel("Happiness Score")

plt.show()
```

Observation

Regions such as Western Europe and North America tend to have higher happiness scores.

Regions such as Sub-Saharan Africa tend to have lower happiness scores.

Interpretation

This regional comparison highlights how economic development, political stability, and social conditions influence happiness across different parts of the world.

 5.6 Correlation with Happiness Score
 Goal

The goal of this analysis is to determine which variables have the strongest relationship with the Happiness Score.

Method

We compute the correlation between the Happiness Score and all other numerical variables in the dataset.
The variables are then sorted from highest to lowest correlation.

This helps identify the factors that most strongly influence happiness levels.

```python
# Correlation with happiness score
corr_happiness = df.corr(numeric_only=True)["Happiness Score"].sort_values(ascending=False)

print(corr_happiness)
```

Quan sát

Kết quả cho thấy các biến số như Kinh tế (GDP bình quân đầu người), Sức khỏe và Gia đình có mối tương quan cao nhất với Điểm Hạnh phúc.

Giải thích

Điều này cho thấy sự thịnh vượng kinh tế, sự hỗ trợ xã hội mạnh mẽ và điều kiện sức khỏe tốt là những yếu tố quan trọng nhất góp phần vào mức độ hạnh phúc cao hơn ở các quốc gia.
